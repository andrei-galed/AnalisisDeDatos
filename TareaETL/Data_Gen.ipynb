```python
import pandas as pd
import random
from faker import Faker

# Configurar Faker para generar nombres en español
fake = Faker('es_ES')

# Número de registros
num_registros = 3547

# Generar datos
data = {
    "IDventa": list(range(1, num_registros + 1)),
    "Numero de Ticket": list(range(1000,num_registros +1000)),
    "Nombre cliente": [fake.name() for val in range(num_registros)],
    "IDVendedor" : [random.randint(1,10) for val in range(num_registros)]
}

# Crear DataFrame
df = pd.DataFrame(data)

# Guardar como archivo CSV
df.to_csv("Clientes.csv", index=False)
```
